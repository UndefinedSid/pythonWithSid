Another method to create DataFrame 

```python
weather_data = [
    {
        "city": "New York",
        "date": "2026-03-04",
        "temperature_c": 8,
        "humidity_percent": 65,
        "condition": "Cloudy",
        "wind_kph": 12
    },
    {
        "city": "London",
        "date": "2026-03-04",
        "temperature_c": 11,
        "humidity_percent": 72,
        "condition": "Rain",
        "wind_kph": 18
    },
    {
        "city": "Tokyo",
        "date": "2026-03-04",
        "temperature_c": 15,
        "humidity_percent": 55,
        "condition": "Sunny",
        "wind_kph": 9
    },
    {
        "city": "Sydney",
        "date": "2026-03-04",
        "temperature_c": 22,
        "humidity_percent": 60,
        "condition": "Partly Cloudy",
        "wind_kph": 14
    }
]
df_1=pd.DataFrame(weather_data)
df_1
```

Another Method for creating a dataframe with given value only in dataset

```python
weather_values = [
    ["New York", "2026-03-04", 8, 65, "Cloudy", 12],
    ["London", "2026-03-04", 11, 72, "Rain", 18],
    ["Tokyo", "2026-03-04", 15, 55, "Sunny", 9],
    ["Sydney", "2026-03-04", 22, 60, "Partly Cloudy", 14]
]
df_2=pd.DataFrame(weather_values)
df_2
```

```python
# Set column names for df_2 (currently 0,1,2,3,4,5)
df_2.columns = ["city", "date", "temperature_c", "humidity_percent", "condition", "wind_kph"]

# Show column names
print("df_1 columns:", df_1.columns.tolist())
print("df_2 columns:", df_2.columns.tolist())
```
